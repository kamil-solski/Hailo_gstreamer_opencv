### Hailo check

```python
try:
    with hailo_platform.VDevice() as vdevice:
        print("Hailo ready")
except hailo_platform.HailoRTException as e:
    print(f"No device: {e}")
```

### Get device name

```python
def check_hailo():
    try:
        devices = hailo_platform.Device.scan()
        if not devices:
            print("No Hailo device found")
            return False
        print(f"Found {len(devices)} device(s): {devices}")

        with hailo_platform.VDevice() as vdevice:
            print("Hailo device is ready")
        return True
    except hailo_platform.HailoRTException as e:
        print(f"Hailo not available: {e}")
        return False

check_hailo()
```
